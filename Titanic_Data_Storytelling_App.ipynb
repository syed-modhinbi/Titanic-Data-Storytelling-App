# Titanic Data Storytelling App

This project explores the Titanic dataset to understand the factors that influenced passenger survival. Through data analysis and visualizations, we identify patterns related to gender, passenger class, age, and fare.


```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("cleaned_titanic.csv")
df.head()
```

```python
df["survived"].value_counts().plot(kind="bar")
plt.title("Survival Count")
plt.show()
```

### Finding 1

A significant number of passengers did not survive the disaster, indicating the severity of the incident.

```python
pd.crosstab(df["sex"], df["survived"]).plot(kind="bar")
plt.title("Gender vs Survival")
plt.show()
```

### Finding 2

The graph shows a strong relationship between gender and survival. Most female passengers survived the Titanic disaster, whereas a large number of male passengers did not survive. This suggests that women were given higher priority during rescue operations.

```python
pd.crosstab(df["pclass"], df["survived"]).plot(kind="bar")
plt.title("Passenger Class vs Survival")
plt.show()
```

### Finding 3

Passengers traveling in higher classes had better chances of survival.

```python
plt.hist(df["age"], bins=20)
plt.title("Age Distribution")
plt.show()
```

### Finding 4

The age distribution shows that most passengers were young adults. A large number of passengers were between 20 and 40 years old, with the highest concentration around 30 years of age.

```python
plt.scatter(df["fare"], df["survived"])
plt.title("Fare vs Survival")
plt.xlabel("Fare")
plt.ylabel("Survived")
plt.show()
```

### Finding 5

Passengers who paid higher fares generally showed higher survival rates.

# Insights and Findings

1. Female passengers had a higher survival rate.
2. First-class passengers were more likely to survive.
3. Most passengers were young adults.
4. Higher ticket fares were associated with higher survival rates.
5. Passenger class and gender strongly influenced survival.

# Conclusion

The Titanic disaster affected passengers differently based on gender, class, and ticket fare. Women and first-class passengers had a greater chance of survival. Data analysis helps reveal these hidden patterns and provides valuable insights into historical events.
